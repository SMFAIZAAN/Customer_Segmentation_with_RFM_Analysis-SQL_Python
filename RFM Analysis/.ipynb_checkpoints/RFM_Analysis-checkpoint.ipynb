```python
df_raw = pd.read_csv(
    'online_retail_II.csv',
    encoding='ISO-8859-1'
)

print(f"✓ Raw data loaded")
print(f"  Shape      : {df_raw.shape[0]:,} rows × {df_raw.shape[1]} columns")
print(f"  Columns    : {df_raw.columns.tolist()}")
print(f"  Date range : {df_raw['InvoiceDate'].min()} → {df_raw['InvoiceDate'].max()}")
print(f"  Unique customers : {df_raw['Customer ID'].nunique():,}")
```

```python
# ─────────────────────────────────────────────
# SECTION 3: EXPLORATORY DATA ANALYSIS
# Understand the data before touching it.
# ─────────────────────────────────────────────

print("=" * 55)
print("SHAPE")
print("=" * 55)
print(f"Rows   : {df_raw.shape[0]:,}")
print(f"Columns: {df_raw.shape[1]}")

print("\n" + "=" * 55)
print("DATA TYPES")
print("=" * 55)
print(df_raw.dtypes)

print("\n" + "=" * 55)
print("MISSING VALUES")
print("=" * 55)
missing = df_raw.isnull().sum()
missing_pct = (missing / len(df_raw) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing,
                           'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

print("\n" + "=" * 55)
print("SAMPLE ROWS")
print("=" * 55)
print(df_raw.head(5))

print("\n" + "=" * 55)
print("NUMERIC SUMMARY")
print("=" * 55)
print(df_raw.describe())

print("\n" + "=" * 55)
print("KEY COUNTS")
print("=" * 55)
print(f"Unique invoices  : {df_raw['Invoice'].nunique():,}")
print(f"Unique customers : {df_raw['Customer ID'].nunique():,}")
print(f"Unique products  : {df_raw['StockCode'].nunique():,}")
print(f"Unique countries : {df_raw['Country'].nunique():,}")
print(f"Cancelled orders : {df_raw['Invoice'].astype(str).str.startswith('C').sum():,}")
print(f"Negative Qty rows: {(df_raw['Quantity'] < 0).sum():,}")
print(f"Zero Price rows  : {(df_raw['Price'] == 0).sum():,}")
```

```python
# ─────────────────────────────────────────────
# SECTION 4: DATA CLEANING
# Every decision is documented with a reason.
# ─────────────────────────────────────────────

df = df_raw.copy()
initial_rows = len(df)
print(f"Starting rows: {initial_rows:,}\n")

# ── Step 1: Standardize column names ──────────
df.columns = (df.columns
                .str.strip()
                .str.replace(' ', '_')
                .str.lower())
print(f"Step 1 — Columns renamed: {df.columns.tolist()}\n")

# ── Step 2: Remove cancellations ──────────────
# In this dataset cancellations appear as
# negative quantities, not 'C' prefix invoices.
before = len(df)
df = df[df['quantity'] > 0]
print(f"Step 2 — Removed cancellations/negatives: -{before - len(df):,} rows")

# ── Step 3: Remove non-product stock codes ─────
bad_codes = ['POST', 'D', 'M', 'BANK CHARGES', 'PADS',
             'DOT', 'CRUK', 'C2', 'S', 'AMAZONFEE']
before = len(df)
df = df[~df['stockcode'].astype(str).str.strip().isin(bad_codes)]
print(f"Step 3 — Removed non-product codes     : -{before - len(df):,} rows")

# ── Step 4: Remove missing Customer IDs ───────
# RFM requires a customer identifier.
# Rows without Customer ID cannot be segmented.
before = len(df)
df = df.dropna(subset=['customer_id'])
print(f"Step 4 — Removed missing CustomerID    : -{before - len(df):,} rows")

# ── Step 5: Remove zero prices ────────────────
# Zero price items are samples or data errors.
before = len(df)
df = df[df['price'] > 0]
print(f"Step 5 — Removed zero/invalid prices   : -{before - len(df):,} rows")

# ── Step 6: Fix data types ────────────────────
df['customer_id'] = df['customer_id'].astype(int)
df['invoicedate'] = pd.to_datetime(df['invoicedate'])
print(f"Step 6 — Data types fixed")

# ── Step 7: Remove price outliers ─────────────
# Items above 99.9th percentile are extreme
# bulk orders that distort monetary scores.
price_cap = df['price'].quantile(0.999)
before = len(df)
df = df[df['price'] <= price_cap]
print(f"Step 7 — Removed price outliers        : -{before - len(df):,} rows  (cap: £{price_cap:.2f})")

# ── Step 8: Create TotalRevenue column ────────
df['total_revenue'] = (df['quantity'] * df['price']).round(2)
print(f"Step 8 — Created total_revenue column")

# ── Step 9: Remove duplicate rows ─────────────
before = len(df)
df = df.drop_duplicates()
print(f"Step 9 — Removed duplicate rows        : -{before - len(df):,} rows")

# ── Final Summary ─────────────────────────────
removed_total = initial_rows - len(df)
print(f"\n{'='*45}")
print(f"CLEANING COMPLETE")
print(f"{'='*45}")
print(f"  Before : {initial_rows:,} rows")
print(f"  After  : {len(df):,} rows")
print(f"  Removed: {removed_total:,} rows ({removed_total/initial_rows*100:.1f}%)")
print(f"  Unique customers : {df['customer_id'].nunique():,}")
print(f"  Unique invoices  : {df['invoice'].nunique():,}")
print(f"  Date range       : {df['invoicedate'].min().date()} → {df['invoicedate'].max().date()}")
print(f"  Total revenue    : £{df['total_revenue'].sum():,.2f}")
```

```python
# ─────────────────────────────────────────────
# SECTION 5: POST-CLEANING VERIFICATION
# ─────────────────────────────────────────────

checks = {
    'Null customer_id'   : df['customer_id'].isnull().sum(),
    'Null invoice'       : df['invoice'].isnull().sum(),
    'Null invoicedate'   : df['invoicedate'].isnull().sum(),
    'Negative quantity'  : (df['quantity'] <= 0).sum(),
    'Negative price'     : (df['price'] <= 0).sum(),
    'Negative revenue'   : (df['total_revenue'] <= 0).sum(),
    'Duplicate rows'     : df.duplicated().sum()
}

print("POST-CLEANING QUALITY CHECKS")
print("─" * 35)
all_passed = True
for check, count in checks.items():
    status = "✓ PASS" if count == 0 else "✗ FAIL"
    if count != 0:
        all_passed = False
    print(f"  {status}  {check}: {count}")

print("─" * 35)
if all_passed:
    print("  All checks passed. Data is clean.")
else:
    print("  Some checks failed. Review before loading.")
```

```python
from urllib.parse import quote_plus

MYSQL_PASSWORD = "Your Password"
MYSQL_USER     = "root"
MYSQL_HOST     = "localhost"
MYSQL_PORT     = "3306"
MYSQL_DB       = "rfm_analysis"

connection_string = (
    f"mysql+mysqlconnector://{MYSQL_USER}:{quote_plus(MYSQL_PASSWORD)}"
    f"@{MYSQL_HOST}:{MYSQL_PORT}/{MYSQL_DB}"
)

engine = create_engine(connection_string)


# Select only the columns MySQL needs
cols_to_load = ['invoice', 'stockcode', 'description',
                'quantity', 'invoicedate', 'price',
                'customer_id', 'country', 'total_revenue']

df_load = df[cols_to_load].copy()

print(f"Loading {len(df_load):,} rows to MySQL...")
print("This may take 1-2 minutes...")

df_load.to_sql(
    name       = 'transactions',
    con        = engine,
    if_exists  = 'replace',   # drops and recreates on re-run
    index      = False,
    chunksize  = 5000
)

print("✓ Data loaded to MySQL successfully.")

# Verify the load
with engine.connect() as conn:
    result = conn.execute(text("SELECT COUNT(*) FROM transactions")).fetchone()
    print(f"✓ MySQL row count: {result[0]:,}")
```

```python
# ─────────────────────────────────────────────
# SECTION 7: PULL RFM DATA FROM MYSQL
# Reads the final segmented view back into Python
# for visualization and analysis.
# ─────────────────────────────────────────────

# Re-use the engine created in Cell 6
# (If restarting notebook: re-run Cell 6 first)

query = "SELECT * FROM v_rfm_segments"
df_segments = pd.read_sql(query, engine)

print(f"✓ RFM data pulled from MySQL")
print(f"  Rows    : {len(df_segments):,}")
print(f"  Columns : {df_segments.columns.tolist()}")
print(f"\nSegment distribution:")
print(df_segments['segment'].value_counts())
```

```python
# ─────────────────────────────────────────────
# SECTION 8: SEGMENT COLOUR MAP & SUMMARY
# ─────────────────────────────────────────────

SEGMENT_COLORS = {
    'Champions'          : '#16a34a',
    'Loyal Customers'    : '#22c55e',
    'Potential Loyalists': '#3b82f6',
    'New Customers'      : '#06b6d4',
    'Promising'          : '#8b5cf6',
    'Need Attention'     : '#f59e0b',
    'About to Sleep'     : '#f97316',
    'At Risk'            : '#ef4444',
    'Cannot Lose Them'   : '#b91c1c',
    'Hibernating'        : '#9ca3af',
    'Lost'               : '#4b5563'
}

# Segment summary table
summary = (
    df_segments
    .groupby('segment')
    .agg(
        customers      = ('customer_id', 'count'),
        total_revenue  = ('monetary',    'sum'),
        avg_revenue    = ('monetary',    'mean'),
        avg_recency    = ('recency_days','mean'),
        avg_frequency  = ('frequency',   'mean')
    )
    .round(2)
    .reset_index()
)

summary['pct_customers'] = (
    summary['customers'] / summary['customers'].sum() * 100
).round(1)

summary['pct_revenue'] = (
    summary['total_revenue'] / summary['total_revenue'].sum() * 100
).round(1)

summary['avg_order_value'] = (
    summary['avg_revenue'] / summary['avg_frequency']
).round(2)

summary = summary.sort_values('total_revenue', ascending=False).reset_index(drop=True)

print("FULL SEGMENT SUMMARY")
print(summary[['segment', 'customers', 'pct_customers',
               'total_revenue', 'pct_revenue',
               'avg_revenue', 'avg_order_value']].to_string(index=False))
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 1: Customer Count per Segment
# ─────────────────────────────────────────────

seg_count = (df_segments['segment']
             .value_counts()
             .reset_index()
             .rename(columns={'index':'segment', 'count':'customers'}))

fig = px.bar(
    seg_count.sort_values('customers', ascending=True),
    x            = 'customers',
    y            = 'segment',
    orientation  = 'h',
    color        = 'segment',
    color_discrete_map = SEGMENT_COLORS,
    title        = 'Customer Count by Segment',
    text         = 'customers',
    labels       = {'customers': 'Number of Customers', 'segment': ''}
)
fig.update_traces(textposition='outside')
fig.update_layout(
    showlegend   = False,
    height       = 550,
    plot_bgcolor = 'white',
    title_font_size = 18
)
fig.show()
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 2: Total Revenue by Segment
# ─────────────────────────────────────────────

rev_by_seg = (
    df_segments
    .groupby('segment')['monetary']
    .sum()
    .reset_index()
    .rename(columns={'monetary': 'total_revenue'})
    .sort_values('total_revenue', ascending=True)
)

rev_by_seg['total_revenue_k'] = (rev_by_seg['total_revenue'] / 1000).round(1)

fig = px.bar(
    rev_by_seg,
    x           = 'total_revenue_k',
    y           = 'segment',
    orientation = 'h',
    color       = 'segment',
    color_discrete_map = SEGMENT_COLORS,
    title       = 'Total Revenue by Segment (£ thousands)',
    text        = 'total_revenue_k',
    labels      = {'total_revenue_k': 'Revenue (£K)', 'segment': ''}
)
fig.update_traces(texttemplate='£%{text:.1f}K', textposition='outside')
fig.update_layout(
    showlegend  = False,
    height      = 550,
    plot_bgcolor= 'white',
    title_font_size = 18
)
fig.show()
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 3: Average RFM Scores Heatmap
# Shows which segments score high/low on each
# dimension. Great chart to discuss in interviews.
# ─────────────────────────────────────────────

heatmap_data = (
    df_segments
    .groupby('segment')[['r_score', 'f_score', 'm_score']]
    .mean()
    .round(2)
    .rename(columns={
        'r_score': 'Recency Score',
        'f_score': 'Frequency Score',
        'm_score': 'Monetary Score'
    })
)

# Sort by total rfm score
heatmap_data = heatmap_data.loc[
    summary['segment'].values
]

fig = px.imshow(
    heatmap_data,
    color_continuous_scale = 'RdYlGn',
    title  = 'Average RFM Scores by Segment',
    aspect = 'auto',
    zmin   = 1,
    zmax   = 4,
    text_auto = True
)
fig.update_layout(
    height = 550,
    title_font_size = 18,
    coloraxis_colorbar_title = 'Score (1–4)'
)
fig.show()
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 4: Recency vs Monetary Scatter
# Coloured by segment. Shows the spread of
# customer value vs. how recently they bought.
# ─────────────────────────────────────────────

# Cap monetary at 99th percentile for readability
monetary_cap = df_segments['monetary'].quantile(0.99)
df_plot = df_segments[df_segments['monetary'] <= monetary_cap].copy()

fig = px.scatter(
    df_plot,
    x           = 'recency_days',
    y           = 'monetary',
    color       = 'segment',
    color_discrete_map = SEGMENT_COLORS,
    opacity     = 0.6,
    title       = 'Recency vs. Lifetime Value by Segment',
    labels      = {
        'recency_days' : 'Days Since Last Purchase (lower = more recent)',
        'monetary'     : 'Total Spend (£)',
        'segment'      : 'Segment'
    },
    hover_data  = ['customer_id', 'frequency', 'rfm_string']
)
fig.update_layout(
    height          = 600,
    plot_bgcolor    = 'white',
    title_font_size = 18
)
fig.update_xaxes(showgrid=True, gridcolor='#f0f0f0')
fig.update_yaxes(showgrid=True, gridcolor='#f0f0f0')
fig.show()
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 5: Pareto — 80/20 Rule
# Shows cumulative revenue by customer quintile.
# This is the most business-impressive chart.
# ─────────────────────────────────────────────

df_segments_sorted = df_segments.sort_values('monetary', ascending=False).reset_index(drop=True)
df_segments_sorted['cumulative_revenue'] = df_segments_sorted['monetary'].cumsum()
df_segments_sorted['cumulative_pct']     = (
    df_segments_sorted['cumulative_revenue'] /
    df_segments_sorted['monetary'].sum() * 100
)
df_segments_sorted['customer_pct'] = (
    (df_segments_sorted.index + 1) /
    len(df_segments_sorted) * 100
)

fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    go.Bar(
        x    = df_segments_sorted['customer_pct'],
        y    = df_segments_sorted['monetary'],
        name = 'Individual Revenue',
        marker_color = 'steelblue',
        opacity = 0.6
    ),
    secondary_y = False
)

fig.add_trace(
    go.Scatter(
        x    = df_segments_sorted['customer_pct'],
        y    = df_segments_sorted['cumulative_pct'],
        name = 'Cumulative Revenue %',
        line = dict(color='#ef4444', width=2.5)
    ),
    secondary_y = True
)

# 80% reference line
fig.add_hline(
    y=80, line_dash='dash', line_color='orange',
    annotation_text='80% Revenue', secondary_y=True
)

fig.update_layout(
    title      = 'Pareto Chart — Revenue Concentration (80/20 Rule)',
    height     = 500,
    plot_bgcolor = 'white',
    title_font_size = 18
)
fig.update_xaxes(title_text='Cumulative % of Customers (ranked by spend)')
fig.update_yaxes(title_text='Individual Order Revenue (£)', secondary_y=False)
fig.update_yaxes(title_text='Cumulative Revenue %', secondary_y=True)
fig.show()

# Print the finding
top20_revenue = df_segments.nlargest(int(len(df_segments)*0.2), 'monetary')['monetary'].sum()
total_revenue = df_segments['monetary'].sum()
print(f"\n80/20 FINDING:")
print(f"Top 20% of customers generate "
      f"{top20_revenue/total_revenue*100:.1f}% of total revenue.")
```

```python
# ─────────────────────────────────────────────
# SECTION 9: BUSINESS QUESTION ANSWERS
# ─────────────────────────────────────────────

total_rev    = df_segments['monetary'].sum()
total_cust   = len(df_segments)

print("=" * 60)
print("Q1: WHO ARE YOUR CHAMPIONS?")
print("=" * 60)
champions = df_segments[df_segments['segment'] == 'Champions']
print(f"  Count                : {len(champions):,} customers ({len(champions)/total_cust*100:.1f}%)")
print(f"  Total Revenue        : £{champions['monetary'].sum():,.2f}")
print(f"  Revenue Share        : {champions['monetary'].sum()/total_rev*100:.1f}% of all revenue")
print(f"  Avg Spend            : £{champions['monetary'].mean():,.2f}")
print(f"  Avg Days Since Buy   : {champions['recency_days'].mean():.0f} days")
print(f"  Avg Orders           : {champions['frequency'].mean():.1f}")

print("\n" + "=" * 60)
print("Q2: WHICH CUSTOMERS ARE AT RISK?")
print("=" * 60)
at_risk_segs = ['At Risk', 'Cannot Lose Them', 'About to Sleep']
at_risk = df_segments[df_segments['segment'].isin(at_risk_segs)]
at_risk_summary = at_risk.groupby('segment').agg(
    count   = ('customer_id', 'count'),
    revenue = ('monetary',    'sum'),
    avg_recency = ('recency_days', 'mean')
).round(2)
print(at_risk_summary)
print(f"\n  Total revenue at risk: £{at_risk['monetary'].sum():,.2f}")

print("\n" + "=" * 60)
print("Q3: DOES THE 80/20 RULE APPLY?")
print("=" * 60)
quintile_size = len(df_segments) // 5
top20   = df_segments.nlargest(quintile_size, 'monetary')['monetary'].sum()
top40   = df_segments.nlargest(quintile_size*2, 'monetary')['monetary'].sum()
print(f"  Top 20% customers → {top20/total_rev*100:.1f}% of revenue")
print(f"  Top 40% customers → {top40/total_rev*100:.1f}% of revenue")
print(f"  Bottom 20% customers → "
      f"{df_segments.nsmallest(quintile_size,'monetary')['monetary'].sum()/total_rev*100:.1f}% of revenue")

print("\n" + "=" * 60)
print("Q5: AVG ORDER VALUE BY SEGMENT")
print("=" * 60)
aov = df_segments.copy()
aov['aov'] = aov['monetary'] / aov['frequency']
aov_summary = (
    aov.groupby('segment')['aov']
    .mean()
    .round(2)
    .sort_values(ascending=False)
    .reset_index()
)
aov_summary.columns = ['Segment', 'Avg Order Value (£)']
print(aov_summary.to_string(index=False))
```

```python
# ─────────────────────────────────────────────
# Q4: CAMPAIGN STRATEGY BY SEGMENT
# This is the "So what?" of the entire project.
# ─────────────────────────────────────────────

strategy = {
    'Champions': {
        'Priority' : '🔴 Critical',
        'Action'   : 'Reward & Retain',
        'Campaign' : 'VIP loyalty program, early access, referral incentive',
        'Channel'  : 'Personalized email, exclusive events'
    },
    'Loyal Customers': {
        'Priority' : '🔴 High',
        'Action'   : 'Upsell & Cross-sell',
        'Campaign' : 'Premium product recommendations, bundle offers',
        'Channel'  : 'Email, targeted ads'
    },
    'Potential Loyalists': {
        'Priority' : '🟡 High',
        'Action'   : 'Nurture to Loyalty',
        'Campaign' : 'Membership program invite, personalized recommendations',
        'Channel'  : 'Email, push notifications'
    },
    'New Customers': {
        'Priority' : '🟡 Medium',
        'Action'   : 'Onboard & Engage',
        'Campaign' : 'Welcome series, product discovery, first-repeat incentive',
        'Channel'  : 'Welcome email sequence'
    },
    'Promising': {
        'Priority' : '🟡 Medium',
        'Action'   : 'Build Relationship',
        'Campaign' : 'Showcase popular products, limited-time offer',
        'Channel'  : 'Email, retargeting ads'
    },
    'Need Attention': {
        'Priority' : '🟡 Medium',
        'Action'   : 'Re-engage',
        'Campaign' : '"We miss you" offer, highlight new arrivals',
        'Channel'  : 'Email with discount'
    },
    'About to Sleep': {
        'Priority' : '🟠 Urgent',
        'Action'   : 'Reactivate Now',
        'Campaign' : 'Time-limited discount, survey to understand drop-off',
        'Channel'  : 'Urgent email + SMS'
    },
    'At Risk': {
        'Priority' : '🔴 Urgent',
        'Action'   : 'Win Back',
        'Campaign' : 'Strong reactivation offer, personal outreach for high-value',
        'Channel'  : 'Email + personal call for top spenders'
    },
    'Cannot Lose Them': {
        'Priority' : '🚨 Critical',
        'Action'   : 'Win Back Immediately',
        'Campaign' : 'Best possible offer, understand pain points, executive outreach',
        'Channel'  : 'Personal phone call + email'
    },
    'Hibernating': {
        'Priority' : '🟢 Low',
        'Action'   : 'Low-cost Re-engagement',
        'Campaign' : 'Occasional newsletter, test with small discount',
        'Channel'  : 'Low-frequency email'
    },
    'Lost': {
        'Priority' : '⚪ Minimal',
        'Action'   : 'Final Attempt or Suppress',
        'Campaign' : 'One last win-back email, then remove from active lists',
        'Channel'  : 'Single email'
    }
}

strategy_df = pd.DataFrame(strategy).T.reset_index()
strategy_df.columns = ['Segment', 'Priority', 'Action', 'Campaign', 'Channel']

# Merge with revenue data
seg_rev = summary[['segment', 'customers', 'total_revenue', 'pct_revenue']]
seg_rev.columns = ['Segment', 'Customers', 'Total Revenue (£)', '% Revenue']

strategy_full = strategy_df.merge(seg_rev, on='Segment', how='left')
strategy_full['Total Revenue (£)'] = strategy_full['Total Revenue (£)'].apply(
    lambda x: f"£{x:,.0f}"
)

print("CAMPAIGN STRATEGY RECOMMENDATIONS BY SEGMENT")
print("=" * 90)
for _, row in strategy_full.iterrows():
    print(f"\n{row['Priority']} | {row['Segment']}")
    print(f"   Customers : {row['Customers']:,}  |  Revenue: {row['Total Revenue (£)']}"
          f"  ({row['% Revenue']}%)")
    print(f"   Action    : {row['Action']}")
    print(f"   Campaign  : {row['Campaign']}")
    print(f"   Channel   : {row['Channel']}")
```

```python
# ─────────────────────────────────────────────
# VISUALIZATION 6: Strategy Summary Table
# Clean table suitable for a report or README.
# ─────────────────────────────────────────────

table_data = strategy_full[[
    'Segment', 'Customers', '% Revenue', 'Action', 'Channel'
]].copy()

fig = go.Figure(data=[go.Table(
    columnwidth = [180, 100, 100, 200, 180],
    header=dict(
        values     = ['<b>Segment</b>', '<b>Customers</b>',
                      '<b>% Revenue</b>', '<b>Action</b>', '<b>Channel</b>'],
        fill_color = '#1e293b',
        font       = dict(color='white', size=12),
        align      = 'left',
        height     = 35
    ),
    cells=dict(
        values     = [table_data[col] for col in table_data.columns],
        fill_color = [['#f8fafc' if i % 2 == 0 else '#ffffff'
                       for i in range(len(table_data))]],
        font       = dict(size=11),
        align      = 'left',
        height     = 30
    )
)])

fig.update_layout(
    title       = 'Customer Segment Strategy Recommendations',
    height      = 460,
    margin      = dict(l=0, r=0, t=40, b=0),
    title_font_size = 16
)
fig.show()
```
